# Ноутбук 2. Проекция — «тень» одного вектора на другой
*(к уроку 04 курса)*

Проекция — последний кирпич перед Грамом–Шмидтом. Идея:

- солнце светит сверху, палка `v` отбрасывает **тень** на землю `e`;
- формула тени: **`тень = (v · e) × e`**, где `e` — единичный вектор;
- **вычесть тень из v → остаётся перпендикуляр**. Это сердце Грама–Шмидта!

Запусти подготовку:

```python
import numpy as np
import numpy.linalg as la
import matplotlib.pyplot as plt   # библиотека для рисования графиков
print("Готово!")
```

## 1. Считаем тень по шагам

```python
v = np.array([3.0, 1.0])          # "палка"
e = np.array([0.6, 0.8])          # "земля" - единичный вектор (проверь: 0.36+0.64=1)

dlina_teni = v @ e                # шаг 1: число - сколько v идёт вдоль e
ten = dlina_teni * e              # шаг 2: сама тень (вектор вдоль e)
perp = v - ten                    # шаг 3: перпендикулярная часть

print("длина тени  :", dlina_teni)
print("тень        :", ten)
print("перпендикуляр:", perp)
print("проверка perp @ e (должен быть ~0):", perp @ e)
```

Число вида `2e-16` — это `0.0000000000000002`, «компьютерный ноль»:
компьютер хранит дроби приближённо, поэтому вместо идеального нуля бывает
микроскопический «мусор». Читай его как ноль.

## 2. А теперь НАРИСУЕМ

Функция ниже рисует векторы стрелками. Не обязательно понимать каждую её строку —
главное запустить и смотреть на картинку:

```python
def risuy_ten(v, e_napravlenie):
    e = e_napravlenie / la.norm(e_napravlenie)   # на всякий случай нормализуем
    ten = (v @ e) * e
    perp = v - ten

    plt.figure(figsize=(6, 6))
    ax = plt.gca()
    ax.axhline(0, color="#ccc"); ax.axvline(0, color="#ccc")

    def strelka(vec, color, name, start=(0, 0)):
        ax.annotate("", xy=(start[0]+vec[0], start[1]+vec[1]), xytext=start,
                    arrowprops=dict(arrowstyle="->", color=color, lw=2.5))
        ax.text(start[0]+vec[0]*1.06, start[1]+vec[1]*1.06, name,
                color=color, fontsize=13, weight="bold")

    strelka(v, "tab:blue", "v")
    strelka(e, "tab:green", "e")
    strelka(ten, "tab:orange", "тень")
    strelka(perp, "tab:purple", "перпендикуляр", start=ten)  # рисуем ИЗ конца тени

    m = max(3, abs(v[0])+1, abs(v[1])+1)
    ax.set_xlim(-m, m); ax.set_ylim(-m, m); ax.set_aspect("equal")
    ax.grid(True, alpha=0.3)
    plt.title("тень = (v·e)e,  перпендикуляр = v - тень")
    plt.show()

risuy_ten(np.array([3.0, 1.0]), np.array([0.6, 0.8]))
```

**Поиграйся с картинкой** — меняй векторы в последней строке и запускай:

- `risuy_ten(np.array([2.0, 3.0]), np.array([1.0, 0.0]))` — тень на ось x;
- сделай v перпендикулярным e — тень исчезнет;
- сделай v вдоль e — исчезнет перпендикуляр.
