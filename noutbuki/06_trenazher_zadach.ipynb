# Ноутбук 6. Тренажёр: бесконечные задачи для закрепления
*(к уроку 11 курса)*

Здесь всё готово — просто пользуйся:

- ячейка «НОВАЯ ЗАДАЧА» придумывает случайную задачу;
- решаешь на бумаге;
- ячейка «ПОКАЗАТЬ РЕШЕНИЕ» печатает пошаговый разбор.

Запусти подготовку:

```python
import numpy as np
import numpy.linalg as la

verySmallNumber = 1e-14

def gsBasis(A):
    B = np.array(A, dtype=np.float64)
    for j in range(B.shape[1]):
        for i in range(j):
            B[:, j] = B[:, j] - B[:, j] @ B[:, i] * B[:, i]
        if la.norm(B[:, j]) > verySmallNumber:
            B[:, j] = B[:, j] / la.norm(B[:, j])
        else:
            B[:, j] = np.zeros_like(B[:, j])
    return B

def novaya_zadacha(razmer=2, ot=-3, do=3, s_lovushkoy=False):
    """Придумывает случайную задачу на Грам-Шмидт."""
    while True:
        A = np.random.randint(ot, do + 1, size=(razmer, razmer)).astype(np.float64)
        if la.norm(A[:, 0]) < verySmallNumber:
            continue                       # первый вектор не должен быть нулём
        if s_lovushkoy and razmer >= 2:
            A[:, -1] = A[:, :-1].sum(axis=1)   # последний = сумма остальных (зависимый!)
        return A

def pokazat_reshenie(A):
    """Печатает решение шаг за шагом, как в уроке 07."""
    B = np.array(A, dtype=np.float64)
    for j in range(B.shape[1]):
        print("-" * 44)
        print("Вектор номер", j + 1, ":", A[:, j])
        for i in range(j):
            d = B[:, j] @ B[:, i]
            print("  тень на e%d: длина %.4f, вычитаем %s" % (i+1, d, np.round(d * B[:, i], 4)))
            B[:, j] = B[:, j] - d * B[:, i]
            print("  осталось:", np.round(B[:, j], 4))
        n = la.norm(B[:, j])
        if n > verySmallNumber:
            B[:, j] = B[:, j] / n
            print("  длина остатка %.4f -> делим: e%d = %s" % (n, j+1, np.round(B[:, j], 4)))
        else:
            B[:, j] = np.zeros_like(B[:, j])
            print("  остаток исчез! вектор ЗАВИСИМЫЙ -> e%d = нули" % (j+1))
    print("=" * 44)
    print("ИТОГ (по столбцам):")
    print(np.round(B, 4))
    return B

print("Тренажёр заряжен!")
```

## 🎲 НОВАЯ ЗАДАЧА — запускай эту ячейку сколько хочешь раз

Каждый запуск — новая случайная задача. Настройки:
`razmer=2` — плоскость (легко на бумаге), `razmer=3` — пространство;
`s_lovushkoy=True` — подсунуть зависимый вектор.

```python
zadacha = novaya_zadacha(razmer=2, ot=-3, do=3, s_lovushkoy=False)

print("Сделай процесс Грама-Шмидта для векторов (это СТОЛБЦЫ матрицы):")
print(zadacha)
for j in range(zadacha.shape[1]):
    print("  v%d = %s" % (j + 1, zadacha[:, j]))
print()
print("Реши на бумаге, ПОТОМ запусти следующую ячейку за решением.")
```

## 🔓 ПОКАЗАТЬ РЕШЕНИЕ — запускай только после того, как решил!

```python
otvet = pokazat_reshenie(zadacha)
```

## 🛠️ Задача «наоборот»: придумай СВОЮ задачу с красивым ответом

Рецепт из урока 11: возьми готовые перпендикулярные векторы и «испорти» их —
собери из них кривые исходники. Ответ будешь знать заранее!

```python
# 1) Возьми перпендикулярную пару (проверь: p @ q = 0):
p = np.array([1.0, 1.0])
q = np.array([1.0, -1.0])
print("p @ q =", p @ q, " <- 0, перпендикулярны")

# 2) Собери "кривые" векторы (меняй множители как хочешь):
v1 = 2 * p                # смотрит вдоль p
v2 = 3 * p + 5 * q        # смесь p и q
print("твоя задача: v1 =", v1, ", v2 =", v2)

# 3) Ты знаешь ответ заранее: e1 = p/|p|, e2 = +-q/|q|. Проверим:
V = np.column_stack([v1, v2])     # склеить векторы в матрицу по столбцам
print("ответ gsBasis:")
print(np.round(gsBasis(V), 4))
print("p нормализованный:", np.round(p / la.norm(p), 4))
print("q нормализованный:", np.round(q / la.norm(q), 4))
```

Заметь: `e2` может получиться со знаком минус — если q-часть в `v2` взята
с отрицательным множителем, Грам–Шмидт развернёт ось в другую сторону.
Оба варианта правильные: перпендикулярность и длина 1 сохраняются.
