# Ноутбук 3. Грам–Шмидт по шагам
*(к урокам 07–08 курса)*

Весь процесс — три действия, повторённые несколько раз:

1. первый вектор → **нормализовать** (получили ось e1);
2. следующий вектор → **вычесть тени** на все готовые оси → **нормализовать** остаток;
3. повторять, пока векторы не кончатся.

Пройдём классический пример из лекций **руками, но в коде** — каждый шаг
отдельной ячейкой, чтобы видеть все промежуточные числа.

```python
import numpy as np
import numpy.linalg as la
import matplotlib.pyplot as plt

v1 = np.array([1.0, 1.0, 1.0])
v2 = np.array([2.0, 0.0, 1.0])
v3 = np.array([3.0, 1.0, -1.0])
print("исходные 'кривые' векторы:")
print("v1 =", v1, " v2 =", v2, " v3 =", v3)
```

## Шаг 1. Нормализуем v1 → e1

```python
e1 = v1 / la.norm(v1)
print("длина v1:", la.norm(v1), " (это корень из 3 =", np.sqrt(3), ")")
print("e1 =", e1)
print("длина e1:", la.norm(e1), " <- стала 1")
```

Сравни с бумажным ответом из урока 07: `e1 = 1/√3 × [1, 1, 1] ≈ [0.577, 0.577, 0.577]` ✓

## Шаг 2. Чистим v2: вычитаем тень на e1

```python
ten = (v2 @ e1) * e1              # тень v2 на ось e1
u2 = v2 - ten                     # остаток - он перпендикулярен e1

print("длина тени (v2 @ e1):", v2 @ e1)
print("тень:", ten)
print("u2 = v2 - тень =", u2)
print("проверка u2 @ e1 (ноль?):", u2 @ e1)
```

Бумага говорила: `u2 = [1, -1, 0]` ✓

## Шаг 3. Нормализуем u2 → e2

```python
e2 = u2 / la.norm(u2)
print("e2 =", e2)      # это 1/sqrt(2) * [1, -1, 0]
```

## Шаг 4. Чистим v3: вычитаем тени на ОБЕ оси e1 и e2

```python
u3 = v3 - (v3 @ e1) * e1          # минус тень на e1
print("после вычитания тени на e1:", u3)

u3 = u3 - (u3 @ e2) * e2          # минус тень на e2 (из того, что осталось!)
print("после вычитания тени на e2:", u3)
```

Бумага: `u3 = [1, 1, -2]` ✓

## Шаг 5. Нормализуем u3 → e3, и финальная проверка

```python
e3 = u3 / la.norm(u3)
print("e3 =", e3)
print()
print("=== ПРОВЕРКА ОТВЕТА (две группы условий из урока 07) ===")
print("длины (должны быть 1):", la.norm(e1), la.norm(e2), la.norm(e3))
print("e1 @ e2 =", e1 @ e2)
print("e1 @ e3 =", e1 @ e3)
print("e2 @ e3 =", e2 @ e3, " <- все три близки к нулю: перпендикулярно!")
```

## Бонус: посмотрим на 2D-пример глазами

В 2D можно нарисовать «до и после». Слева кривые v1, v2 — справа ровные e1, e2:

```python
def gs_2d_kartinka(v1, v2):
    # Грам-Шмидт для двух 2D-векторов
    e1 = v1 / la.norm(v1)
    u2 = v2 - (v2 @ e1) * e1
    e2 = u2 / la.norm(u2)

    fig, axes = plt.subplots(1, 2, figsize=(11, 5.5))
    for ax, pary, title in [
        (axes[0], [(v1, "tab:blue", "v1"), (v2, "tab:orange", "v2")], "ДО: кривые"),
        (axes[1], [(e1, "tab:green", "e1"), (e2, "tab:purple", "e2")], "ПОСЛЕ: перпендикулярные, длины 1"),
    ]:
        ax.axhline(0, color="#ccc"); ax.axvline(0, color="#ccc")
        for vec, color, name in pary:
            ax.annotate("", xy=tuple(vec), xytext=(0, 0),
                        arrowprops=dict(arrowstyle="->", color=color, lw=2.5))
            ax.text(vec[0]*1.08, vec[1]*1.08, name, color=color, fontsize=13, weight="bold")
        m = max(3.2, abs(v1).max()+1, abs(v2).max()+1)
        ax.set_xlim(-m, m); ax.set_ylim(-m, m); ax.set_aspect("equal")
        ax.grid(True, alpha=0.3); ax.set_title(title)
    plt.show()
    print("e1 =", e1, "  e2 =", e2, "  e1@e2 =", e1 @ e2)

gs_2d_kartinka(np.array([3.0, 4.0]), np.array([2.0, 1.0]))
```
