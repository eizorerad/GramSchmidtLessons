## 1. Подключаем NumPy

## 2. Векторы — стрелки из чисел

Вектор `[2, 3]` — это инструкция «иди на 2 вправо и на 3 вверх».
Складываются векторы по позициям, умножение на число растягивает стрелку:

```python
v = np.array([2, 3])            # вектор v
w = np.array([1, 1])            # вектор w

print("v + w =", v + w)         # сложение по позициям: [3 4]
print("v - w =", v - w)         # вычитание по позициям: [1 2]
print("2 * v =", 2 * v)         # растянули в 2 раза: [4 6]
print("v / 2 =", v / 2)         # сжали в 2 раза: [1. 1.5]
```

**Поиграйся:** поменяй числа в `v` и `w` в ячейке выше, предскажи результат,
запусти и проверь себя. Это главный способ учиться в ноутбуках!

## 3. Длина вектора — теорема Пифагора

Длина `[3, 4]` — это `√(3² + 4²) = √25 = 5`. В NumPy её считает `la.norm`:

```python
v = np.array([3, 4])
print("длина v:", la.norm(v))                        # 5.0
print("проверим Пифагора сами:", np.sqrt(3**2 + 4**2))   # тоже 5.0
```

## 4. Нормализация — делаем длину равной 1

Рецепт: **раздели вектор на его длину**. Направление сохранится, длина станет 1.
Векторы длиной 1 называются *единичными* и обозначаются буквой `e`:

```python
v = np.array([3, 4], dtype=np.float64)   # float64 = храним как дробные числа
e = v / la.norm(v)                       # делим на длину
print("e =", e)                          # [0.6 0.8]
print("длина e:", la.norm(e))            # 1.0 - получилось!
```

## 5. Скалярное произведение: значок `@`

Перемножаем по позициям и складываем — получается **одно число**:

`[1, 2] · [3, 4] = 1×3 + 2×4 = 11`

Главное правило: **результат 0 ⇔ векторы перпендикулярны (угол 90°)**.

⚠️ Не путай `@` и `*`:

```python
u = np.array([1, 2])
v = np.array([3, 4])

print("u @ v =", u @ v)     # скалярное произведение: ЧИСЛО 11
print("u * v =", u * v)     # поэлементное умножение: ВЕКТОР [3 8]

# проверка перпендикулярности:
a = np.array([2, 1])
b = np.array([-1, 2])
print("a @ b =", a @ b, " <- ноль, значит a и b ПЕРПЕНДИКУЛЯРНЫ")
```

## 6. Векторы в столбцах матрицы

В задачах Грама–Шмидта несколько векторов упаковывают в матрицу **как столбцы**.
Запись `A[:, 0]` читается «все строки, столбец 0» — то есть первый вектор:

```python
A = np.array([[1, 2, 3],
              [1, 0, 1],
              [1, 1, -1]], dtype=np.float64)

print("v1 =", A[:, 0])     # первый столбец  [1. 1. 1.]
print("v2 =", A[:, 1])     # второй столбец  [2. 0. 1.]
print("v3 =", A[:, 2])     # третий столбец  [3. 1. -1.]
print()
print("а A[0] - это СТРОКА:", A[0])   # не перепутай!
```
